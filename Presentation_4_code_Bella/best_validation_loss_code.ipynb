```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load predictions
df = pd.read_csv("best_val_loss_boxplot_ready.csv")
```

# Best Validation Loss
For each of the five splits, plot a boxplot of the best validation loss obtained across the 10 runs.

```python
# Step 1: create mapping
mapping = {
    "split_1_data_splits_new_01": "Split 1",
    "split_2_data_splits_new_02": "Split 2",
    "split_3_data_splits_new_03": "Split 3",
    "split_4_data_splits_new_04": "Split 4",
    "split_5_data_splits_new_05": "Split 5"
}

# Step 2: apply mapping
df["split_label"] = df["split_id"].map(mapping)

# Step 3: plot using the new column
plt.figure(figsize=(10,6))

sns.boxplot(x="split_label", y="best_val_loss", data=df)
sns.stripplot(x="split_label", y="best_val_loss", data=df, jitter=True, alpha=0.6)

plt.xlabel("Data Split")
plt.ylabel("Best Validation Loss")
plt.title("Best Validation Loss Across Seeds (w/ Individual Runs)")

plt.show()
```
